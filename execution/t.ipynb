```python
# Import the executor
from .executor import HyperliquidExecutor

# Create an instance
executor = HyperliquidExecutor(testnet=True)

# Example usage:
try:
    # Get market data
    markets = executor.get_markets()
    print("Available Markets:", markets)
    
    # Get current positions
    positions = executor.get_positions()
    print("\nCurrent Positions:", positions)
    
    # Example: Open a long position in ETH
    long_result = executor.open_long(
        coin="ETH",
        size=0.01,  # 0.01 ETH
        slippage=0.005  # 0.5% slippage
    )
    print("\nLong Position Result:", long_result)
    
    # Example: Close the position
    close_result = executor.close_position(coin="ETH")
    print("\nClose Position Result:", close_result)
    
    # Or close all positions at once
    # all_closes = executor.close_all_positions()
    # print("\nClose All Positions Result:", all_closes)
    
except Exception as e:
    print("Error:", str(e))



```

```python
from hyperliquid.exchange import Exchange
from eth_account import Account
import os
from dotenv import load_dotenv

load_dotenv()

# Get credentials from .env
pk = os.getenv("PK")
address = os.getenv("ADDRESS")

# Create wallet
wallet = Account.from_key(pk)

# Initialize exchange
exchange = Exchange(
    wallet=wallet,
    base_url=None,
    account_address=address
)

# Test connection and get account info
try:
    # Get available markets and prices
    markets = exchange.info.all_mids()
    print("Available Markets:", markets)
    
    # Get user positions using the correct API call
    positions = exchange.info.user_state(address)
    print("\nCurrent Positions:", positions)
    
    # Let's try to place a test order for BTC
    # Using market_open method for a market order
    # test_order = exchange.market_open(
    #     name="BTC",      # Trading pair
    #     is_buy=True,     # Buy order
    #     sz=0.001,        # Size in BTC
    #     slippage=0.01    # 1% slippage tolerance
    # )
    # print("\nTest Order Result:", test_order)
    
except Exception as e:
  print("Error:", str(e))
```
